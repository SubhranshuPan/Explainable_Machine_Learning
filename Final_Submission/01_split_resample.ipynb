```python
import glob
from pathlib import Path

import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
# FIX: removed `from os import pa` (typo, crashed the notebook) and `from scipy import *` (unused)
sns.set()

# FIX: all paths are relative to the project root, so the notebook runs the same
# whether it is started from the project root or from the notebooks/ folder.
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
RAW_DIR = ROOT / "mitbih"            # raw MIT-BIH files (.dat/.hea/.atr)
DATA_DIR = ROOT / "data"             # generated CSV files
RECORD_DIR = DATA_DIR / "records"    # one CSV per record and channel
FIG_DIR = ROOT / "figures"           # figures used in the report
for d in (DATA_DIR, RECORD_DIR, FIG_DIR):
    d.mkdir(parents=True, exist_ok=True)
```

```python
length = 277
data = np.empty(shape=[0, length])

all_data = glob.glob(str(DATA_DIR / 'all_data.csv'))

for j in all_data:
    print('Loading ', j)
    csvrows = np.loadtxt(j, delimiter=',')
    data = np.append(data, csvrows, axis=0)

print(data.shape)
```

# Test Train Split

* The beat and patient hold-out methods were used for training and testing, split into 75-25%, and 90-10% respectively.


```python
from sklearn.model_selection import train_test_split
train, test = train_test_split(data,test_size=0.25,random_state=48)
print(train.shape)
print(test.shape)
```

# Data Resampling

* We see an abundance of ’N’ beats, while all the other beat classes do not even pass the 10,000 threshold. 
* To address the imbalance between the classes in the MIT-BIH dataset, we used the resample technique by Sci-kit Learn.
* This resampling method uses bootstrap method which estimates statistics on a data population by sampling a dataset with replacement through iteration.



# Beat Split and Resample

* For up-sampling and down-sampling, the sample value was calculated by taking the mean values of the total number of beats of the abnormal classes. 
* We can see beats with their respective annotation classes from both ECG channels.


```python
from sklearn.utils import resample

N = train[train[:,-2]==1.0]
L = train[train[:,-2]==2.0]
R = train[train[:,-2]==3.0]
V = train[train[:,-2]==4.0]
A = train[train[:,-2]==5.0]
F = train[train[:,-2]==6.0]
f = train[train[:,-2]==7.0]
I = train[train[:,-2]==8.0]

seed=42
np.random.seed(seed)
def downsample(arr, n, seed):
    downsampled = resample(arr,replace=False,n_samples=n, random_state=seed)
    return downsampled

def upsample(arr, n, seed):
    upsampled = resample(arr,replace=True,n_samples=n,random_state=seed)
    return upsampled

all_class = [N,L,R,V,A,F,f,I]
abn_class = [L,R,V,A,F,f,I]

mean_val = np.mean([len(i) for i in abn_class], dtype= int)
train_sampled = []

for i in all_class:
    if i.shape[0]> mean_val:
        i = downsample(i,mean_val,seed)
    elif i.shape[0]< mean_val:
        i = upsample(i, mean_val,seed)
    train_sampled.append(i)
    
train_values = np.concatenate(train_sampled)
np.random.shuffle(train_values)
train_sampled = train_values

with open(DATA_DIR / 'train_beats.csv', "wb") as fin:
    np.savetxt(fin, train_sampled, delimiter=",", fmt='%f')

with open(DATA_DIR / 'test_beats.csv', "wb") as fin:
    np.savetxt(fin, test, delimiter=",", fmt='%f')
```

# Patient Split and Resample

* Patients that were left out for testing were: 104, 208, 113, 210, 119 (the remaining 43 patients are used for training).
* __FIX:__ the original loop rebuilt the list of training rows once per held-out patient and concatenated the five lists.
  Each list only excluded *one* patient, so the training set contained every beat ~5 times, **including the beats of the
  other four test patients** (data leakage). This is why the original plot showed 25,789 beats per class.
  A boolean mask now selects test rows (`isin`) and training rows (everything else) exactly once.

```python
train_patients = np.empty(shape=[0, length])
test_patients = np.empty(shape=[0, length])
patient_number = [104,208,113,210,119]

# FIX: select each row exactly once (see the markdown cell above)
is_test_patient = np.isin(data[:, -1], patient_number)
test_patients = data[is_test_patient]
train_patients = data[~is_test_patient]

# Sanity checks: no patient is in both sets and no row is lost or duplicated
assert set(np.unique(train_patients[:, -1])).isdisjoint(patient_number)
assert len(train_patients) + len(test_patients) == len(data)
print('Train patients:', len(np.unique(train_patients[:, -1])), 'records,', train_patients.shape)
print('Test patients: ', np.unique(test_patients[:, -1]).astype(int), test_patients.shape)

from sklearn.utils import resample

N = train_patients[train_patients[:,-2]==1.0]
L = train_patients[train_patients[:,-2]==2.0]
R = train_patients[train_patients[:,-2]==3.0]
V = train_patients[train_patients[:,-2]==4.0]
A = train_patients[train_patients[:,-2]==5.0]
F = train_patients[train_patients[:,-2]==6.0]
f = train_patients[train_patients[:,-2]==7.0]
I = train_patients[train_patients[:,-2]==8.0]

seed=42
np.random.seed(seed)
def downsample(arr, n, seed):
    downsampled = resample(arr,replace=False,n_samples=n, random_state=seed)
    return downsampled

def upsample(arr, n, seed):
    upsampled = resample(arr,replace=True,n_samples=n,random_state=seed)
    return upsampled

all_class = [N,L,R,V,A,F,f,I]
abn_class = [L,R,V,A,F,f,I]

mean_val = np.mean([len(i) for i in abn_class], dtype= int)
train_sampled = []

for i in all_class:
    if i.shape[0]> mean_val:
        i = downsample(i,mean_val,seed)
    elif i.shape[0]< mean_val:
        i = upsample(i, mean_val,seed)
    train_sampled.append(i)
    
train_sampled = np.concatenate(train_sampled)
np.random.shuffle(train_sampled)
train_sampled_all = train_sampled

with open(DATA_DIR / 'train_patients.csv', "wb") as fin:
    np.savetxt(fin, train_sampled_all, delimiter=",", fmt='%f')

with open(DATA_DIR / 'test_patients.csv', "wb") as fin:
    np.savetxt(fin, test_patients, delimiter=",", fmt='%f')
```

# Function to plot counts

```python
def plot_counts(df,name,color,fname=None):
  counts = df[275].value_counts()
  plt.figure(figsize=(8,4))
  feature_names = ['N','L','R','V','A','F','f','/']
  barplt = plt.bar(counts.index, counts.values, alpha=0.8, color=color)
  plt.title(name)
  plt.ylabel('Number of Occurrences', fontsize=12)
  plt.xlabel('Beat Category', fontsize=12)
  plt.xticks(ticks=[1,2,3,4,5,6,7,8],labels=feature_names)
  for bar in barplt:
    yval = bar.get_height()
    plt.text(bar.get_x()+.2, yval+600, yval)
  if fname:
    plt.savefig(FIG_DIR / fname, dpi=200, bbox_inches='tight')
  plt.show()
```
